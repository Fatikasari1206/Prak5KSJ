```python
import numpy as np

# Fungsi untuk mengubah huruf ke angka
def text_to_numbers(text):
    return [ord(char.upper()) - ord('A') for char in text if char.isalpha()]

# Fungsi untuk mengubah angka ke huruf
def numbers_to_text(numbers):
    return ''.join([chr(num % 26 + ord('A')) for num in numbers])

# Fungsi enkripsi Hill Cipher
def hill_encrypt(plain_text, key_matrix):
    key_size = key_matrix.shape[0]
    numbers = text_to_numbers(plain_text)

    # Padding jika panjang teks tidak kelipatan ukuran kunci
    while len(numbers) % key_size != 0:
        numbers.append(23)  # 'X'

    cipher = []
    for i in range(0, len(numbers), key_size):
        chunk = np.array(numbers[i:i+key_size])
        result = np.dot(key_matrix, chunk) % 26
        cipher.extend(result)

    return numbers_to_text(cipher)

# Contoh penggunaan
if __name__ == "__main__":
    key = np.array([[3, 3], [2, 5]])  # Contoh matriks kunci 2x2
    plaintext = "AKU SAYANG AKU"

    cipher_text = hill_encrypt(plaintext, key)
    print(f"Plaintext: {plaintext}")
    print(f"Ciphertext: {cipher_text}")
```
